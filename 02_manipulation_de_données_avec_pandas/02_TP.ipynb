# 62. Le réflexe à développer

Ne pensez pas :

> "Quelle fonction Pandas dois-je utiliser ?"

Pensez plutôt :

> **"Quelle question est-ce que je cherche à résoudre ?"**

Puis décomposez-la.

Par exemple :

> "Quel est le montant moyen des commandes françaises de plus de 100 € ?"

devient :

```text
1. Partir du DataFrame
       ↓
2. Garder France
       ↓
3. Garder amount > 100
       ↓
4. Sélectionner amount
       ↓
5. Calculer mean()
```

Puis :

````python
df[
    (df["country"] == "France")
    & (df["amount"] > 100)
]["amount"].mean()
````
