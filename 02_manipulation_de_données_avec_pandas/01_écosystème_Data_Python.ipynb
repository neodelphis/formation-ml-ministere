# 2. Manipuler des données avec Python "pur"

Imaginons que nous ayons quelques clients :

````python
customers = [
    {"name": "Alice", "age": 32, "city": "Paris"},
    {"name": "Bob", "age": 41, "city": "Lyon"},
    {"name": "Charlie", "age": 27, "city": "Paris"},
]
````

Afficher les clients :

````python
for customer in customers:
    print(customer)
````

Filtrer les clients de Paris :

````python
for customer in customers:
    if customer["city"] == "Paris":
        print(customer["name"])
````

Calculer l'âge moyen :

````python
ages = [customer["age"] for customer in customers]

sum(ages) / len(ages)
````

Ça fonctionne.

Mais imaginons maintenant que nous ayons :

* 100 000 clients ;
* 50 colonnes ;
* des valeurs manquantes ;
* plusieurs fichiers ;
* des dates ;
* des catégories ;
* des calculs statistiques.

Le code Python "pur" devient rapidement difficile à maintenir.


# 3. NumPy : le calcul numérique

## Le problème

NumPy est spécialisé dans le **calcul numérique efficace**.

Son objet central est le `ndarray`.

````python
import numpy as np
````

Créons un tableau :

````python
ages = np.array([32, 41, 27, 35, 29])

ages
````

Calculons quelques statistiques :

````python
ages.mean()
````

````python
ages.min()
````

````python
ages.max()
````

````python
ages.std()
````

## Vectorisation

Une des forces de NumPy est la possibilité d'effectuer des opérations sur l'ensemble des données.

````python
ages + 1
````

Plutôt que :

````python
new_ages = []

for age in ages:
    new_ages.append(age + 1)

new_ages
````

NumPy permet donc de travailler efficacement avec des **tableaux numériques**.

# 6. DataFrame : penser "table"

Un `DataFrame` peut être vu comme une **table de données**.

```text
              colonnes
                  ↓
        ┌─────────┬─────┬────────┐
        │  name   │ age │  city  │
        ├─────────┼─────┼────────┤
ligne → │ Alice   │ 32  │ Paris  │
        │ Bob     │ 41  │ Lyon   │
        │ Charlie │ 27  │ Paris  │
        └─────────┴─────┴────────┘
             ↑
           index
```

Chaque colonne représente généralement une **variable**.

Chaque ligne représente généralement une **observation**.

## 7.1 Series

Une `Series` correspond à une colonne de données.

````python
ages = pd.Series([32, 41, 27, 35])

ages
````

Une Series possède :

* des valeurs ;
* un index ;
* éventuellement un nom.

````python
ages.name = "age"

ages
````

## 7.2 DataFrame

Un `DataFrame` est constitué de plusieurs Series partageant le même index.

```text
             DataFrame
                 │
       ┌─────────┼─────────┐
       ↓         ↓         ↓
     Series    Series    Series
      name       age      city
```

On peut donc voir :

````python
df["age"]
````

comme une Series.

# 11. Pandas et les opérations vectorisées

C'est une notion fondamentale.

Nous pouvons créer une nouvelle colonne :

````python
df["salary_k"] = df["salary"] / 1000

df
````

Nous venons d'appliquer une opération à **toute une colonne**.

Pas besoin de boucle :

````python
# ❌ Pas nécessaire

for i in range(len(df)):
    df.loc[i, "salary_k"] = df.loc[i, "salary"] / 1000
````

On préfère :

````python
# ✅ Pandas

df["salary_k"] = df["salary"] / 1000
````

# 14. Premier filtre

Imaginons que nous voulions uniquement les personnes de plus de 30 ans.

````python
df["age"] > 30
````

Pandas nous retourne une série de booléens :

```text
0     True
1     True
2    False
3     True
4    False
```

Nous pouvons utiliser ce résultat pour filtrer le DataFrame :

````python
df[df["age"] > 30]
````

# 15. Un premier aperçu du raisonnement Pandas

Cette expression :

````python
df[df["age"] > 30]
````

peut être lue comme :

> "Donne-moi les lignes de `df` pour lesquelles `age > 30` est vrai."

C'est une idée fondamentale de Pandas :

```text
DataFrame
    │
    │ condition
    ↓
Masque booléen
    │
    ↓
Sous-ensemble
```

Nous passerons beaucoup de temps sur cette mécanique dans le prochain module.

