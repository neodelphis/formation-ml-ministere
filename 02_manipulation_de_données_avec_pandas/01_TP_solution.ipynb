# 18. 🧪 Mini-TP — Prise en main

Nous allons utiliser le DataFrame suivant :

````python
import pandas as pd
````

````python
data = {
    "name": [
        "Alice", "Bob", "Charlie",
        "David", "Emma", "Frank"
    ],
    "age": [32, 41, 27, 35, 29, 52],
    "city": [
        "Paris", "Lyon", "Paris",
        "Marseille", "Lyon", "Paris"
    ],
    "salary": [
        42000, 58000, 39000,
        51000, 45000, 72000
    ]
}

df = pd.DataFrame(data)

df
````

### Questions

**2. Quels sont les noms des colonnes ?**

````python
df.columns
````

**3. Quels sont les types des colonnes ?**

````python
df.info()
````

````python
df.dtypes
````

**4. Affichez uniquement les colonnes `name` et `salary`.**

````python
df[["name", "salary"]]
````

**5. Quel est le salaire moyen ?**

````python
df.salary.mean()
````

**6. Quels sont les employés de plus de 30 ans ?**

````python
df[df["age"] > 30]
````

**7. Créez une colonne `salary_k` contenant le salaire en milliers d'euros.**

````python
df["salary_k"] = df.salary / 1000
df
````


```text

      ↓
```

```text
      ↓

```
