# Arbres de décision avec variables continues

## 1. Ajout d'une colonne température

Nous allons conserver exactement le même problème, mais ajouter une variable continue : **la température**.

| Jour | Météo   | Vent   | Température | Jouer ? |
| ---: | ------- | ------ | ----------: | ------- |
|    1 | Soleil  | Faible |          28 | Oui     |
|    2 | Soleil  | Faible |          30 | Oui     |
|    3 | Soleil  | Fort   |          32 | Non     |
|    4 | Soleil  | Fort   |          34 | Non     |
|    5 | Nuageux | Faible |          24 | Oui     |
|    6 | Nuageux | Fort   |          26 | Oui     |
|    7 | Nuageux | Fort   |          27 | Oui     |
|    8 | Pluie   | Faible |          18 | Non     |
|    9 | Pluie   | Faible |          20 | Non     |
|   10 | Pluie   | Fort   |          22 | Oui     |

La variable $Température$ est **continue**. Contrairement à $Météo$, elle ne possède pas un petit nombre de modalités prédéfinies.


## 2. Séparation selon un seuil

Pour une variable continue, l'arbre va créer des questions binaires du type :

$$Température \leq seuil ?$$

Par exemple :

```text
             Température ≤ 25 ?
                 /        \
               Oui        Non
```


> **Pour une variable continue, l'arbre cherche le seuil qui produit la meilleure séparation.** 

L'évaluation de la qualité de cette séparation se fait grace au calcul du gain sur l'indice de Gini.

## 3. Quels seuils faut-il tester ?

Supposons que nous ayons les températures :

$$18,\ 20,\ 22,\ 24,\ 26,\ 27,\ 28,\ 30,\ 32,\ 34$$

L'algorithme commence par **trier les valeurs**.

Puis il va considérer les seuils situés entre deux valeurs consécutives.

Par exemple :

```text
18    20    22    24    26    27    28    30    32    34
 |-----|-----|-----|-----|-----|-----|-----|-----|-----|
    ↑     ↑     ↑     ↑     ↑     ↑     ↑     ↑     ↑
```

Les seuils candidats sont les milieux :

$$19,\ 21,\ 23,\ 25,\ 26.5,\ 27.5,\ 29,\ 31,\ 33$$

Testons un premier seuil : 25 °C

Prenons :

$$\boxed{Température \leq 25}$$

### Groupe gauche

$$18,20,22,24$$

| Jour | Température | Jouer ? |
| ---: | ----------: | ------- |
|    8 |          18 | Non     |
|    9 |          20 | Non     |
|   10 |          22 | Oui     |
|    5 |          24 | Oui     |

Donc :

* 2 Oui
* 2 Non

Le Gini vaut :

$$Gini_{gauche}=0.5$$

### Groupe droit

$$26,27,28,30,32,34$$

Nous avons :

* 4 Oui
* 2 Non

Donc :

$$Gini_{droit}=1-\left(\frac46\right)^2-\left(\frac26\right)^2$$

$$=1-\frac49-\frac19$$

$$=\frac49$$

Donc :

$$Gini_{droit}\approx0.444$$


### Calcul du Gini du split

Nous avons :

* 4 observations à gauche ;
* 6 à droite.

Donc :

$$Gini_{split}=\frac4{10}(0.5)+\frac6{10}(0.444)$$

$$=0.2+0.2664$$

$$\boxed{Gini_{split}\approx0.466}$$

le Gini initial était :

$$0.48$$

Nous sommes passés à environ :

$$0.466$$

L'amélioration est faible.

## 4. Choix du meilleur candidat

L'algorithme peut donc faire quelque chose qui ressemble à ceci :

| Seuil | Gini du split |
| ----: | ------------: |
|    19 |           ... |
|    21 |           ... |
|    23 |           ... |
|    25 |         0.466 |
|  26,5 |           ... |
|  27,5 |         0.466 |
|    29 |     **0.419** |
|    31 |           ... |
|    33 |           ... |

Il va calculer le Gini pour **tous les seuils candidats**, puis prendre celui qui donne le Gini minimal.

> **Avec une variable continue, le problème de recherche du meilleur attribut devient un problème de recherche du meilleur seuil.**



## 5. Algorithme

Ce que fait l'algorithme :

1. calculer le gain avec `Météo` ;
2. calculer le gain avec `Vent` ;
3. calculer le gain avec meilleur seuil possible pour `Température` ;
4. comparer tous les résultats ;
5. prendre le meilleur.

Donc, la question posée est:

> **« Quelle séparation produit la plus forte réduction d'impureté ? »**

C'est ce qui est choisit à chaque noeud en prenant en considération les colonnes restantes.
