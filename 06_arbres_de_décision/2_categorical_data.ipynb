# Arbres de décision avec variables qualitatives

(`categorical variable`, variable catégorielle, variable qualitative)

## Jeu de données `tennis` 

Imaginons que nous voulions prédire si une personne va **jouer au tennis**.

Nous avons 10 observations :

| Jour | Météo   | Vent   | Jouer ? |
| ---- | ------- | ------ | ------- |
| 1    | Soleil  | Faible | Oui     |
| 2    | Soleil  | Faible | Oui     |
| 3    | Soleil  | Fort   | Non     |
| 4    | Soleil  | Fort   | Non     |
| 5    | Nuageux | Faible | Oui     |
| 6    | Nuageux | Fort   | Oui     |
| 7    | Nuageux | Fort   | Oui     |
| 8    | Pluie   | Faible | Non     |
| 9    | Pluie   | Faible | Non     |
| 10   | Pluie   | Fort   | Oui     |



## 1. Le principe général

Un arbre de décision cherche à transformer progressivement un ensemble d'observations mélangées en sous-ensembles de plus en plus **purs**.

Ici, notre variable cible est :

> **Jouer ?** → Oui / Non

Au départ, nous avons :

* 6 personnes qui jouent ;
* 4 personnes qui ne jouent pas.

L'algorithme va chercher une question du type :

> « Quelle variable permet de séparer au mieux les Oui et les Non ? »

Dans notre cas, les variables candidates sont :

* **Météo**
* **Vent**

L'algorithme va donc essayer différentes séparations et calculer leur qualité avec l'indice de Gini.


## 2. L'indice de Gini

Pour un nœud contenant plusieurs classes, l'indice de Gini est :

$$Gini = 1-\sum_{k=1}^{K}p_k^2$$

où $p_k$ est la proportion d'observations appartenant à la classe (k).

Nous avons seulement deux classes :

* Oui
* Non

Donc :

$$Gini = 1-p_{Oui}^2-p_{Non}^2$$

### Exemple très simple

Imaginons un nœud contenant :

| Jouer ? | Nombre |
| ------- | -----: |
| Oui     |      6 |
| Non     |      4 |

Alors :

$$p_{Oui}=\frac{6}{10}=0.6$$

et

$$p_{Non}=\frac{4}{10}=0.4$$

Donc :

$$Gini = 1-0.6^2-0.4^2$$

$$=1-0.36-0.16$$

$$=\boxed{0.48}$$

Notre ensemble initial a donc un **indice de Gini de 0,48**.

## 3. Que signifie le Gini ?

Le Gini mesure en quelque sorte le **mélange des classes**.

### Nœud parfaitement pur

Si nous avons :

| Jouer ? | Nombre |
| ------- | -----: |
| Oui     |     10 |
| Non     |      0 |

alors :

$$Gini=1-1^2-0^2=0$$

Donc :

$$\boxed{Gini=0}$$

Le nœud est parfaitement pur.


### Nœud très mélangé

Si nous avons :

| Jouer ? | Nombre |
| ------- | -----: |
| Oui     |      5 |
| Non     |      5 |

alors :

$$Gini=1-0.5^2-0.5^2$$

$$=0.5$$

Donc :

$$\boxed{Gini=0.5}$$

Pour deux classes, c'est le maximum.

> **Plus le Gini est proche de 0, plus le nœud est pur.**

## 4. Notre nœud racine

Reprenons les données :

| Jour | Météo   | Vent   | Jouer ? |
| ---: | ------- | ------ | ------- |
|    1 | Soleil  | Faible | Oui     |
|    2 | Soleil  | Faible | Oui     |
|    3 | Soleil  | Fort   | Non     |
|    4 | Soleil  | Fort   | Non     |
|    5 | Nuageux | Faible | Oui     |
|    6 | Nuageux | Fort   | Oui     |
|    7 | Nuageux | Fort   | Oui     |
|    8 | Pluie   | Faible | Non     |
|    9 | Pluie   | Faible | Non     |
|   10 | Pluie   | Fort   | Oui     |

Au départ :

$$6\ Oui,\quad 4\ Non$$

et donc :

$$\boxed{Gini_{racine}=0.48}$$

Maintenant, nous devons décider :

> **Quelle variable utiliser pour effectuer la première séparation ?**

Nous allons tester **Météo** puis **Vent**.

## 5. Premier candidat : Météo

La variable Météo possède trois valeurs :

* Soleil
* Nuageux
* Pluie

Une première séparation pourrait donc produire trois branches :

```text
             Météo
           /   |    \
      Soleil Nuageux Pluie
```

Nous devons calculer le Gini de chacun des trois nouveaux nœuds.

---

### 5.1. Branche Soleil

Les observations correspondantes sont :

| Jour | Météo  | Vent   | Jouer ? |
| ---: | ------ | ------ | ------- |
|    1 | Soleil | Faible | Oui     |
|    2 | Soleil | Faible | Oui     |
|    3 | Soleil | Fort   | Non     |
|    4 | Soleil | Fort   | Non     |

Nous avons :

* 2 Oui
* 2 Non

Donc :

$$p_{Oui}=\frac{2}{4}=0.5$$

$$p_{Non}=\frac{2}{4}=0.5$$

Le Gini vaut :

$$Gini_{Soleil}=1-0.5^2-0.5^2$$

$$=\boxed{0.5}$$

Ce n'est pas très pur.

---

### 5.2. Branche Nuageux

Les observations sont :

| Jour | Météo   | Vent   | Jouer ? |
| ---: | ------- | ------ | ------- |
|    5 | Nuageux | Faible | Oui     |
|    6 | Nuageux | Fort   | Oui     |
|    7 | Nuageux | Fort   | Oui     |

Nous avons :

* 3 Oui
* 0 Non

Donc :

$$Gini_{Nuageux}=1-1^2-0^2$$

$$=\boxed{0}$$

Excellent !

Cette branche est **parfaitement pure**.

---

### 5.3. Branche Pluie

Les observations sont :

| Jour | Météo | Vent   | Jouer ? |
| ---: | ----- | ------ | ------- |
|    8 | Pluie | Faible | Non     |
|    9 | Pluie | Faible | Non     |
|   10 | Pluie | Fort   | Oui     |

Nous avons :

* 1 Oui
* 2 Non

Donc :

$$p_{Oui}=\frac13$$

$$p_{Non}=\frac23$$

Le Gini vaut :

$$Gini_{Pluie}=1-\left(\frac13\right)^2-\left(\frac23\right)^2$$

$$=1-\frac19-\frac49$$

$$=1-\frac59$$

$$=\boxed{\frac49\approx0.444}$$

## 6. Gini de la séparation « Météo »

>**moyenne pondérée** pour estimer l'indice de Gini de cette séparation

Comme les branches n'ont pas le même nombre d'observations, on fait une une **moyenne pondérée**.

- La branche Soleil contient 4 observations.
- Nuageux: 3.
- Pluie: 3.


La formule est :

$$Gini_{split}=\sum_j\frac{N_j}{N}Gini_j$$

* $N$ = nombre total d'observations ;
* $N_j$ = nombre d'observations dans la branche (j).

Donc :

$$Gini_{Météo}=\frac{4}{10}(0.5)+\frac{3}{10}(0)+\frac{3}{10}(0.444)$$

$$=0.2+0+0.1333$$

Donc :

$$\boxed{Gini_{Météo}\approx0.333}$$

La séparation selon Météo fait donc passer le Gini de :

$$0.48 \rightarrow 0.333$$

C'est une amélioration importante.

## 7. Deuxième candidat : Vent

Maintenant, faisons exactement la même chose avec **Vent**.

La variable Vent possède deux valeurs :

* Faible
* Fort

Nous obtenons :

```text
              Vent
             /    \
         Faible    Fort
```

---

### 7.1. Branche Vent = Faible

Regardons les observations :

| Jour | Météo   | Vent   | Jouer ? |
| ---: | ------- | ------ | ------- |
|    1 | Soleil  | Faible | Oui     |
|    2 | Soleil  | Faible | Oui     |
|    5 | Nuageux | Faible | Oui     |
|    8 | Pluie   | Faible | Non     |
|    9 | Pluie   | Faible | Non     |

Nous avons :

* 3 Oui
* 2 Non

Donc :

$$p_{Oui}=\frac35$$

$$p_{Non}=\frac25$$


$$Gini_{Faible}=1-\left(\frac35\right)^2-\left(\frac25\right)^2$$

$$=1-\frac9{25}-\frac4{25}$$

$$=1-\frac{13}{25}$$

$$=\boxed{0.48}$$


### 7.2. Branche Vent = Fort

Les observations sont :

| Jour | Météo   | Vent | Jouer ? |
| ---: | ------- | ---- | ------- |
|    3 | Soleil  | Fort | Non     |
|    4 | Soleil  | Fort | Non     |
|    6 | Nuageux | Fort | Oui     |
|    7 | Nuageux | Fort | Oui     |
|   10 | Pluie   | Fort | Oui     |

Nous avons :

* 3 Oui
* 2 Non

C'est exactement la même distribution.

$$\boxed{Gini_{Fort}=0.48}$$


### 7.3. Gini de la séparation selon Vent

Les deux branches contiennent 5 observations.

$$Gini_{Vent}=\frac5{10}(0.48)+\frac5{10}(0.48)$$

$$=0.24+0.24$$

$$\boxed{Gini_{Vent}=0.48}$$

## 8. Comparaison des deux variables

Nous avons maintenant :

| Variable | Gini après séparation |
| -------- | --------------------: |
| Météo    |             **0.333** |
| Vent     |                 0.480 |

L'objectif est de **minimiser** le Gini après séparation.

Donc :

$$\boxed{\text{Météo est choisie comme racine}}$$

Notre arbre commence donc ainsi :

```text
                    Météo
                /     |      \
            Soleil  Nuageux   Pluie
```

Et nous savons déjà quelque chose d'intéressant : la branche **Nuageux** est pure = **feuille**.

Nous pouvons donc immédiatement mettre :

```mermaid
graph TD
    %% Nœuds de décision (losanges, fond orange)
    Meteo{"Météo"}
    Int1{"?"}
    Int2{"?"}

    %% Feuilles (Résultats, ellipses, fond vert)
    Oui1(["Oui"])

    %% Connexions avec étiquettes (labels)
    Meteo -- "Soleil" --> Int1
    Meteo -- "Nuageux" --> Oui1
    Meteo -- "Pluie" --> Int2

    %% Définition des styles personnalisés
    classDef decision fill:#ffcc80,stroke:#e65100,stroke-width:2px,color:#000
    classDef resultOui fill:#a5d6a7,stroke:#2e7d32,stroke-width:2px,color:#000
    classDef resultNon fill:#ef9a9a,stroke:#c62828,stroke-width:2px,color:#000

    %% Application des styles
    class Meteo,Int1,Int2 decision
    class Oui1 resultOui
```

Mais l'algorithme n'est pas terminé.

## 9. On recommence sur la branche Soleil

C'est un point **fondamental pour comprendre les arbres de décision**.

L'algorithme ne repart pas de zéro avec les 10 observations.

Il travaille maintenant uniquement avec les observations qui sont arrivées dans ce nœud.

Pour Soleil, nous avons :

| Jour | Vent   | Jouer ? |
| ---: | ------ | ------- |
|    1 | Faible | Oui     |
|    2 | Faible | Oui     |
|    3 | Fort   | Non     |
|    4 | Fort   | Non     |

Le nœud contient :

$$2\ Oui,\quad2\ Non$$

Son Gini est :

$$Gini=0.5$$

Il nous reste une variable utilisable :

> **Vent**

Nous pouvons donc essayer :

```text
Soleil
  |
  +---- Vent = Faible
  |
  +---- Vent = Fort
```

---

### 9.1. Soleil + Vent faible

Nous avons les jours 1 et 2 :

| Vent   | Jouer ? |
| ------ | ------- |
| Faible | Oui     |
| Faible | Oui     |

Donc :

$$2\ Oui,\quad0\ Non$$

et :

$$\boxed{Gini=0}$$


### 9.2. Soleil + Vent fort

Nous avons les jours 3 et 4 :

| Vent | Jouer ? |
| ---- | ------- |
| Fort | Non     |
| Fort | Non     |

Donc :

$$0\ Oui,\quad2\ Non$$

et :

$$\boxed{Gini=0}$$

La séparation est donc parfaite :

$$Gini_{Vent|Soleil}=\frac24(0)+\frac24(0)$$

$$=\boxed{0}$$

Nous obtenons :

```mermaid
graph TD
    %% Nœuds de décision (losanges, fond orange)
    Meteo{"Météo"}
    Vent{"Vent"}
    Interrogation{"?"}

    %% Feuilles (Résultats, ellipses, fond vert/rouge)
    Oui1(["Oui"])
    Oui2(["Oui"])
    Non1(["Non"])

    %% Connexions avec étiquettes (labels)
    Meteo -- "Soleil" --> Vent
    Meteo -- "Nuageux" --> Oui2
    Meteo -- "Pluie" --> Interrogation

    Vent -- "Faible" --> Oui1
    Vent -- "Fort" --> Non1

    %% Définition des styles personnalisés
    classDef decision fill:#ffcc80,stroke:#e65100,stroke-width:2px,color:#000
    classDef resultOui fill:#a5d6a7,stroke:#2e7d32,stroke-width:2px,color:#000
    classDef resultNon fill:#ef9a9a,stroke:#c62828,stroke-width:2px,color:#000

    %% Application des styles
    class Meteo,Vent,Interrogation decision
    class Oui1,Oui2 resultOui
    class Non1 resultNon
```

## 10. Et maintenant la branche Pluie

Il reste :

| Jour | Vent   | Jouer ? |
| ---: | ------ | ------- |
|    8 | Faible | Non     |
|    9 | Faible | Non     |
|   10 | Fort   | Oui     |

Nous avons :

$$1\ Oui,\quad2\ Non$$

avec :

$$Gini=\frac49\approx0.444$$

Il reste encore la variable Vent.

### Vent faible

Jours 8 et 9 :

$$0\ Oui,\quad2\ Non$$

Donc :

$$Gini=0$$

### Vent fort

Jour 10 :

$$1\ Oui,\quad0\ Non$$

Donc :

$$Gini=0$$

La séparation est donc parfaite.

## 10. Notre arbre final

On obtient :

```mermaid
graph TD
    %% Nœuds de décision (losanges)
    Meteo{"Météo"}
    Vent1{"Vent"}
    Vent2{"Vent"}

    %% Feuilles (Résultats)
    Oui1(["Oui"])
    Non1(["Non"])
    Oui2(["Oui"])
    Non2(["Non"])
    Oui3(["Oui"])

    %% Connexions avec étiquettes
    Meteo -- "Soleil" --> Vent1
    Meteo -- "Nuageux" --> Oui2
    Meteo -- "Pluie" --> Vent2

    Vent1 -- "Faible" --> Oui1
    Vent1 -- "Fort" --> Non1

    Vent2 -- "Faible" --> Non2
    Vent2 -- "Fort" --> Oui3

    %% Définition des styles personnalisés
    classDef decision fill:#ffcc80,stroke:#e65100,stroke-width:2px,color:#000
    classDef resultOui fill:#a5d6a7,stroke:#2e7d32,stroke-width:2px,color:#000
    classDef resultNon fill:#ef9a9a,stroke:#c62828,stroke-width:2px,color:#000

    %% Application des styles
    class Meteo,Vent1,Vent2 decision
    class Oui1,Oui2,Oui3 resultOui
    class Non1,Non2 resultNon
```

L'arbre classifie donc parfaitement les **10 observations d'apprentissage**.

## 11. Gain de Gini

Jusqu'ici, nous avons comparé directement les Gini après séparation.

On peut aussi exprimer l'amélioration obtenue par une séparation.

On appelle souvent cela la **réduction d'impureté** :

$$Gain = Gini_{parent}-Gini_{split}$$

Pour la racine :

### Météo

$$Gain_{Météo}=0.48-0.333$$

$$\boxed{Gain_{Météo}\approx0.147}$$

### Vent

$$Gain_{Vent}=0.48-0.48$$

$$\boxed{Gain_{Vent}=0}$$

Donc Météo apporte une réduction d'impureté de 0,147 tandis que Vent n'apporte aucune amélioration.

On choisit donc Météo.

## 12. Fonctionnement de l'algorithme.

À chaque nœud :

### Étape 1 — Calculer l'impureté du nœud

Ici :

$$Gini=1-p_{Oui}^2-p_{Non}^2$$

### Étape 2 — Essayer les différentes variables

Par exemple :

* Météo
* Vent

### Étape 3 — Pour chaque variable, créer les sous-groupes

Par exemple :

```text
Météo
├── Soleil
├── Nuageux
└── Pluie
```

### Étape 4 — Calculer le Gini de chaque sous-groupe

### Étape 5 — Calculer le Gini pondéré de la séparation

$$Gini_{split}=\sum_j\frac{N_j}{N}Gini_j$$

### Étape 6 — Choisir la meilleure séparation

Celle qui donne le **plus petit Gini pondéré**.

Équivalent :

> choisir celle qui produit le **plus grand gain d'impureté**.

### Étape 7 — Recommencer récursivement

On applique exactement la même procédure à chaque nœud qui n'est pas encore pur.

Si un noeud est pur (Toutes les observations appartiennent à la même classe,
comme « Nuageux »), il devient immédiatement une feuille.


# 13. Comment l'arbre prédit-il une nouvelle observation ?

Imaginons :

> Météo = Pluie  
> Vent = Faible

On descend dans l'arbre :

```mermaid
graph TD
    %% Nœuds de décision
    Meteo{"Météo"}
    Vent1{"Vent"}
    Vent2{"Vent"}

    %% Feuilles (Résultats)
    Oui1(["Oui"])
    Non1(["Non"])
    Oui2(["Oui"])
    Non2(["Non"])
    Oui3(["Oui"])

    %% Connexions (L'ordre est important pour numéroter les liens de 0 à 6)
    Meteo -- "Soleil" --> Vent1      %% Lien 0
    Meteo -- "Nuageux" --> Oui2     %% Lien 1
    Meteo -- "Pluie" --> Vent2      %% Lien 2 (Chemin emprunté)
    
    Vent1 -- "Faible" --> Oui1      %% Lien 3
    Vent1 -- "Fort" --> Non1        %% Lien 4
    
    Vent2 -- "Faible" --> Non2      %% Lien 5 (Chemin emprunté)
    Vent2 -- "Fort" --> Oui3        %% Lien 6

    %% Définition des styles standards
    classDef decision fill:#ffcc80,stroke:#e65100,stroke-width:2px,color:#000
    classDef resultOui fill:#a5d6a7,stroke:#2e7d32,stroke-width:2px,color:#000
    classDef resultNon fill:#ef9a9a,stroke:#c62828,stroke-width:2px,color:#000

    %% Styles pour le chemin mis en évidence (bordure bleue épaisse)
    classDef pathDecision fill:#ffcc80,stroke:#1565c0,stroke-width:4px,color:#000
    classDef pathResult fill:#ef9a9a,stroke:#1565c0,stroke-width:4px,color:#000

    %% Application des styles standards aux nœuds inactifs
    class Vent1 decision
    class Oui1,Oui2,Oui3 resultOui
    class Non1 resultNon

    %% Application des styles mis en évidence au chemin actif
    class Meteo,Vent2 pathDecision
    class Non2 pathResult

    %% Mise en évidence des flèches empruntées (les liens 2 et 5 en bleu épais)
    linkStyle 2,5 stroke:#1565c0,stroke-width:4px,color:#1565c0
```

Donc :

$$\boxed{\text{Prédiction = Non}}$$
