# 15. Comprendre le critère de Gini

Vous avez peut-être remarqué que chaque nœud de l'arbre contient une information appelée :

```text
gini
```

Le critère de Gini mesure approximativement le degré de **mélange des classes** dans un nœud.

Pour deux classes, il peut être calculé avec :

\[
Gini = 1 - p_1^2 - p_2^2
\]

où :

- \(p_1\) est la proportion de la première classe ;
- \(p_2\) est la proportion de la deuxième classe.

## Exemple

Imaginons un nœud contenant :

- 10 observations `oui` ;
- 0 observation `non`.

Alors :

\[
p_{oui}=1
\]

\[
p_{non}=0
\]

Donc :

\[
Gini = 1 - 1^2 - 0^2 = 0
\]

Le nœud est parfaitement pur.

Imaginons maintenant :

- 5 observations `oui` ;
- 5 observations `non`.

Alors :

\[
p_{oui}=0.5
\]

\[
p_{non}=0.5
\]

Donc :

\[
Gini = 1 - 0.5^2 - 0.5^2
\]

\[
Gini = 0.5
\]

Le nœud est beaucoup plus mélangé.

# 20. Contrôler la profondeur de l'arbre

Un arbre trop profond peut apprendre très précisément les données d'entraînement.

Il risque alors de mémoriser des situations particulières plutôt que d'apprendre des règles générales.

Nous allons limiter sa profondeur.

Créez un nouvel arbre avec :

````python
max_depth=3
````

````python
decision_tree = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
````

Reconstruisez la pipeline.

````python
model_depth_3 = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', decision_tree)
    ]
)
````

Entraînez le modèle.

````python
model_depth_3.fit(X_train, y_train)
````

Évaluez-le sur les données d'entraînement et de test.

````python
train_predictions = model_depth_3.predict(X_train)
test_predictions = model_depth_3.predict(X_test)

train_accuracy = accuracy_score(y_train, train_predictions)
test_accuracy = accuracy_score(y_test, test_predictions)

print(f'Accuracy entraînement : {train_accuracy:.2%}')
print(f'Accuracy test : {test_accuracy:.2%}')
````

# 21. Comparer plusieurs profondeurs

Nous allons maintenant tester différentes valeurs de `max_depth`.

````python
depths = [1, 2, 3, 4, 5, 7, 10, None]

results = []

for depth in depths:
    decision_tree = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )

    current_model = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('classifier', decision_tree)
        ]
    )

    current_model.fit(X_train, y_train)

    train_predictions = current_model.predict(X_train)
    test_predictions = current_model.predict(X_test)

    results.append({
        'max_depth': depth,
        'train_accuracy': accuracy_score(
            y_train,
            train_predictions
        ),
        'test_accuracy': accuracy_score(
            y_test,
            test_predictions
        )
    })
````

Transformez les résultats en DataFrame.

````python
results_df = pd.DataFrame(results)

results_df
````

# 22. Visualiser les performances

Tracez les performances du modèle en fonction de la profondeur.

````python
plt.figure(figsize=(10, 6))

plt.plot(
    results_df['max_depth'].astype(str),
    results_df['train_accuracy'],
    marker='o',
    label='Entraînement'
)

plt.plot(
    results_df['max_depth'].astype(str),
    results_df['test_accuracy'],
    marker='o',
    label='Test'
)

plt.xlabel('Profondeur maximale')
plt.ylabel('Accuracy')
plt.title('Performance du modèle selon la profondeur')
plt.legend()
plt.grid()

plt.show()
````

# 30. Bilan du TP

Vous avez maintenant construit votre premier modèle de classification basé sur un arbre de décision.

Vous avez découvert que le modèle :

1. cherche des séparations dans les données ;
2. choisit les séparations permettant d'obtenir des groupes plus homogènes ;
3. construit progressivement une succession de règles ;
4. aboutit à des feuilles contenant les prédictions ;
5. peut devenir trop complexe ;
6. doit donc être contrôlé pour éviter le surapprentissage.

Une représentation simplifiée du fonctionnement est :

```text
                    Données
                       │
                       ▼
             Quelle séparation ?
                       │
              ┌────────┴────────┐
              │                 │
           Condition          Condition
             vraie              fausse
              │                 │
              ▼                 ▼
          Sous-groupe       Sous-groupe
              │                 │
              ▼                 ▼
       Nouvelle règle      Nouvelle règle
              │                 │
             ...               ...
              │                 │
              └────────┬────────┘
                       ▼
                    Prédiction
```

Dans le prochain cours, nous verrons comment **combiner plusieurs arbres de décision** afin d'obtenir un modèle généralement plus robuste : les **forêts aléatoires**.
